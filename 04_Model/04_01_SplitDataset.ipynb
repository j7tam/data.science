```python
import pandas as pd
from sklearn.model_selection import train_test_split
from torch.utils.data import DataLoader
```

```python
import sys
import json
from pathlib import Path

ROOT = Path.cwd().resolve().parents[0]
sys.path.append(str(ROOT))

from Utils.project_utils import *
from Utils.transforms import *
from Utils.MushroomDataset import *
```

```python
PROJECT_ROOT = get_project_root()
CLASSES_DIR = PROJECT_ROOT / "Classes"
DATA_DIR = PROJECT_ROOT / "Data"

device = get_device()
```

```python
df_balanced = pd.read_csv(DATA_DIR / "mushroomBalanced.csv")
```

#### Preparing Train, Val, Test Datasets & Loaders

```python
train, temp = train_test_split(df_balanced, test_size=0.2, random_state=42, stratify=df_balanced['class'])
test, val = train_test_split(temp, test_size=0.5, random_state=42, stratify=temp['class'])

CLASSES = sorted(train["class"].unique().tolist())

train.shape, val.shape, test.shape
```

```python
SPLIT_DIR = DATA_DIR / "splits"
SPLIT_DIR.mkdir(parents=True, exist_ok=True)

train.to_csv(SPLIT_DIR / "train.csv", index=False)
val.to_csv(SPLIT_DIR / "val.csv", index=False)
test.to_csv(SPLIT_DIR / "test.csv", index=False)
```

```python
train_class_to_idx = {c: i for i, c in enumerate(CLASSES)}
train_class_to_idx
```

```python
with open(SPLIT_DIR / "class_to_idx.json", "w") as f:
    json.dump(train_class_to_idx, f)
```
